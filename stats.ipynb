```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

df=pd.read_csv("results/single_var/data.csv")
df.head()
df
```

```python
df.info()
```

```python
dx_list=df["DX"].unique()
dx_list[0]
```

```python
d1=df[df["DX"]==dx_list[0]].copy()
```

```python
sns.countplot(data=d1,x="abs_theta",hue="D_diff")
```

```python
plt.figure(figsize=(9, 5))

d1["D_diff"] = d1["D_diff"].astype("category")
sns.boxplot(data=d1, x="D_diff", y="abs_theta")
plt.tight_layout()
```
